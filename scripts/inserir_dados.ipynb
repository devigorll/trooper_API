# **Código:**

```python
# Conectando ao banco de dados

import pyodbc

conn = pyodbc.connect(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=localhost;"
    "DATABASE=trooperAPI_db;"
    "Trusted_Connection=yes;"
)

cursor = conn.cursor()
```
